# Question 02: [ Marks 20 ]

Manual Gradient Calculation

Given:

z = w*x + b <br>
y_pred = sigmoid(z) <br>
L = −[ y log(y_pred) + (1−y) log(1−y_pred) ] <br>

a) Write dL/dy_pred <br>
b) Write dy_pred/dz <br>
c) Show that dL/dz = y_pred − y <br>

## Write Answer 02:

(a) Compute $\frac{dL}{dy_{pred}}$

The loss function is:
$$
L = -(ylog(y_{pred}) + (1-y)log(1-y_{pred}))
$$

Differentiating w.r.t $y_{pred}:$
$$
\frac{dL}{dy_{pred}} = - (\frac{y}{y_{pred}} - \frac{1-y}{1-y_{pred}})
$$
So,
$$
\frac{dL}{dy_{pred}} = - \frac{y}{y_{pred}} + \frac{1-y}{1-y_{pred}}
$$

(b) Compute $\frac{dy_{pred}}{dz}$

Since $y_{pred} = \sigma(z) = \frac{1}{1+e^{-z}}$, we know the derivative of sigmoid:
$$
\frac{dy_{pred}}{dz} = y_{pred}(1 - y_{pred})
$$

(c) Compute $\frac{dL}{dz}$

Using the chain rule:
$$
\frac{dL}{dz} = \frac{dL}{dy_{pred}}. \frac{dy_{pred}}{dz}
$$

Substitute $\frac{dL}{dy_{pred}}$ and $\frac{dy_{pred}}{dz}:$
$$
\frac{dL}{dz} = (-\frac{y}{y_{pred}} + \frac{1-y}{1-y_{pred}}) ⋅ (y_{pred}(1-y_{pred}))
$$

Simplify step by step:
1. Multiply terms:
$$
-\frac{y}{y_{pred}} \cdot y_{pred}(1-y_{pred}) = -y(1-y_{pred})
$$

$$
\frac{1-y}{1-y_{pred}} \cdot y_{pred}(1-y_{pred}) = (1-y)y_{pred}
$$

2. Combine both:

$$
\frac{dL}{dz} = -y(1-y_{pred}) + (1-y)y_{pred}
$$

$$
\frac{dL}{dz}  = -y + yy_{pred} + y_{pred} - yy_{pred}
$$

$$
\frac{dL}{dz} = y_{pred} - y
$$

Hence proved:

$$
\frac{dL}{dz} = y_{pred} - y
$$

## Write Answer 03:

(a) Computation Graph

The computation graph shows how the value of $x$ flows through $y_1$ and $y_2$ to $L$:

(b) Compute $\frac{dL}{dx}$ manually

Step 1: $L$ in terms of $x$:

$$
L = y_1 + y_2 = x^2 + 4x
$$

Step 2: Differentiate w.r.t $x$:

$$
\frac{dL}{dx} = \frac{d(x^2)}{dx} + \frac{d(4x)}{dx}
$$

$$
\frac{dL}{dx} = 2x + 4
$$

Step 3: Substitute $x=3$

$$
\frac{dL}{dx} = 2(3) + 4 = 6 + 4 = 10
$$

So, $\frac{dL}{dx} = 10$

(c) Compute using Autograd

Autograd automatically calculates derivatives using the computation graph

Step 1: Define $x$ as a variable with gradient tracking enabled:

```python
import torch
x = torch.tensor(3.0, requires_grad=True)
```

Step 2: Compute $y_1, y_2, L:$

```python
y1 = x**2
y2 = 4*x
L = y1 + y2
```

Step 3. Call `.backward()` on the output $L$ to compute gradients:

```python
L.backward()
```

Step 4: The gradient of $L$ w.r.t $x$ is sorted in:

```python
print(x.grad) # Output: 10
```

Explanation:
* Autograd automatically builds a computation graph in the background
* When `.backward()` is called, it propagates gradients backward using chain rule
* No manual differentiation is needed

# Question 04: [ Marks 15 ]

Create a scalar tensor x = 3.0 with gradient tracking enabled. <br>
Compute y = x² + 4x. <br>
Perform backpropagation. <br>
Print the gradient of x. <br>

Write the full PyTorch code yourself. <br>

## Write Answer 04:

```python
# here I create a tensor x = 3.0 and enable gradient tracking
x = torch.tensor(3.0, requires_grad=True)

# now I calculate y = x² + 4x
y = x**2 + 4*x

# using backward() to compute the gradient
y.backward()

# printing the gradient of x
print("Gradient of x:", x.grad)
```

Explanation:
The function is $y = x^2 + 4x$. Its derivative is $\frac{dy}{dx} = 2x +4$. At $x=3$, the gradient is $10$. PyTorch compute this automatically using autograd.

# Question 05: [ Marks 10 ]

Create two tensors:
w = 2.0 (requires_grad=True) <br>
b = 1.0 (requires_grad=True) <br>

Let input x = 3.0. <br>
Compute z = w*x + b. <br>
Use Sigmoid to get y_pred.
 <br>

Define Binary Cross Entropy manually (without built-in loss). <br>
Compute gradients using backward(). <br>
Print gradients of w and b. <br>

Write full code.

## Write Answer 05:

```python
# here I define weights and bias with gradient tracking
w = torch.tensor(2.0, requires_grad=True)
b = torch.tensor(1.0, requires_grad=True)

# input value
x = torch.tensor(3.0)

# compute z = w*x +b
z = w*x + b

# apply sigmoid to get prediction
y_pred = 1 / (1 + torch.exp(-z))

# assume true label (for example y = 1)
y = torch.tensor(1.0)

# manually define binary cross entropy loss
loss = -(y * torch.log(y_pred) + (1 - y) * torch.log(1 - y_pred))

# perform backpropagation
loss.backward()

# print gradients of w and b
print("Gradient of w:", w.grad)
print("Gradient of b:", b.grad)
```

Explanation: The gradients are very small because the predicted value is already very close to the true label. The negative sign indicates that the model will slightly decrease the values of $w$ and $b$ to reduce the loss further.

Create tensor w = 2.0 (requires_grad=True). <br>

Print w.grad. <br>

Do not clear gradients. <br>

## Write Answer 07:



```python
# here I create a tensor w = 2.0 with gradient tracking
w = torch.tensor(2.0, requires_grad=True)

# first loss: L1 = w^2
L1 = w**2

# backward for first loss
L1.backward()

# second loss: L2 = 3*w
L2 = 3 * w

# backward for second loss (without clearing gradient)
L2.backward()

# printing the gradient of w
print("Gradient of w:", w.grad)
```

What I Observed: I observed that the gradients are accumulated when `.backward()` is called multiple times without clearing gradients. PyTorch does not overwrite the previous gradient, instead it adds the new gradient to the existing one. That is why the final gradient becomes the sum of both gradients.

So:
* First backward → gradient = 4
* Second backward → gradient = 3
* Final stored gradient = 7
