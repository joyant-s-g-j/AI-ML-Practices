# Question 01: [ Marks 10 ]

What is a perceptron?
Explain its three main components and their roles.

## Write Answer 01:

A perceptron is the basic unit of a neural network. It is inspired by how a biological neuron works. A perceptron receives numerical input values process them, and produces an output based on a mathematical function. It is mainly used for simple classification tasks.
The perceptron has three main components:

1. Inputs and Weights: The inputs are numerical values (such as 0.5, 2.3, -1 etc.) that represent features of the data. Each input has an associated weight. The role of weights is to show how important each input is.Inputs with larger weights have more effect on the final result.

2. Weighted Sum and Activation Function: The perceptron multiplies each input by its weight and then adds them together. This is called the weighted sum. After this, an activation function is applied. The activation function decides the output value. In a simple perceptron, a step or threshold function is often used, but the calculation itself is based on numerical values.

3. The output of a perceptron is also a numerical value. In many cases, this output is used to represent a class (for example, 0 or 1), but mathematically it comes from numerical computation, not just an on/off switch.


# Question 02: [ Marks 10 ]

What is a decision boundary in a perceptron?
Explain how the equation
w₁x₁ + w₂x₂ + b = 0
represents a decision boundary geometrically.

## Write Answer 02:

A decision boundary in a perceptron is an imaginary line that divides the input space into two regions, and the perceptron uses this line to decide which class a data point belongs to. It separates data points of different classes based on the perceptron's output. In a perceptron, the decision is made using the equation:
$$ w_1 x_1 + w_2 x_2 + b = 0 $$

Explanation:
* Here, $x_1$ and $x_2$ are input features (coordinates of a data point).
* $w_1$ and $w_2$ are weights that control the direction and orientation of the boundary.
* $b$ is the bias, which shifts the boundary from the origin.

This equation represents a straight line in a 2D space. All points that satisfy the equation lie exactly on the decision boundary.

* If a point satisfies
$$ w_1 x_1 + w_2 x_2 + b > 0 $$
it lies on one side of the line and is classified as one class.

* If a point satisfies
$$ w_1 x_1 + w_2 x_2 + b < 0 $$
it lies on the other side of the line and is classified as the other class.

The weights ($w_1$, $w_2$) determine the slope of the line, while the bias (b) determines the position of the line. Thus, the perceptron uses this straight line as a decision boundary to separate input data into two classes.

# Question 03: [ Marks 15 ]

Why is the perceptron called a linear classifier?
Explain your answer using the concept of linear separability and logical gate examples such as AND, OR, and XOR.

## Write Answer 03:

A perceptron is called a linear classifier because it separates input data using a linear decision boundary, such as a straight line in two dimensions or a plane in higher dimensions. This means the perceptron can correctly classify data only when the classes can be separated by a straight line. This property is known as linear separability.

**Linear separability**: Data is said to be linearly separable if we can draw a single straight line that separates data points of different classes without any overlap. The perceptron works well for such data because it makes decisions using a linear equation of the form:
$$ w_1 x_1 + w_2 x_2 + b = 0 $$
This equation always produces a straight-line decision boundary, which is why the perceptron is called a linear classifier.

**Logical gate example**
* **AND gate**: The AND gate outputs 1 only when both inputs are 1. The input-output pairs of an AND gate are linearly separable. A single straight line can separate the output 1 case from the output 0 cases. Therefore, a perceptron can easily learn and implement the AND gate.

* **OR gate**: The OR gate outputs 1 when at least one input is 1. Like the AND gate, its input-output pairs are also linearly separable. A straight line can separate the 1 outputs from the 0 output. Hence, a perceptron can also represent the OR gate.

* **XOR gate**: The XOR gate outputs 1 when the inputs are different. In this case, the data points are not linearly separable. No single straight line can separate the 1 outputs from the 0 outputs. Because a perceptron can only form linear decision boundaries, it cannot correctly learn the XOR gate.


# Question 04: [ Marks 15 ]

Write the perceptron weight update rule and explain it intuitively.
Why are the weights updated only when the prediction is incorrect?

## Write Answer 04:

**Perceptron Weight Update Rule**: The perceptron updates its weights using the formula:

$$
w_{i,j}^{\text{(next step)}} = w_{i, j} + \eta(y_j - \hat{y}_j)x_i  
$$

* $w_{i,j}$ is the connection weight between the $i^{\text{th}}$ input and the $j^{\text{th}}$ output
* $\eta$ is the learning rate
* $y_j$ is the actual target value
* $\hat{y}_j$ is the current predicted output of the perceptron
* $x_i$ is the input value

The bias can be updated similarly as:
$$
b \gets b + \eta(y_j - \hat{y}_j)
$$

**Intuitive Explanation**
1. The perceptron makes a prediction using the current weights.
2. If the prediction is correct $(y_j = \hat{y}_j),$ nothing changes.
3. If the prediction is wrong $(y_j \neq \hat{y}_j),$ the weights are updated:
* We increase weights for inputs that should have contributed more to the correct output.
* We decrease weights for inputs that caused the wrong output.

This way, the perceptron learns from its mistakes and gradually improves its predictions.


**Why we update weights only when prediction is wrong?**
* Updating weights when the prediction is already correct is unnecessary.
* It would unnecessarily disturb the model, making it less stable.
* Focusing only on mistakes ensures faster and more efficient learning.





# Question 05: [ Marks 10 ]

Why does a single-layer perceptron fail to solve the XOR problem?
What does this limitation tell us about the need for multilayer neural networks?

## Write Answer 05:

A single-layer perceptron fails to solve the XOR problem because XOR is not linearly separable.
* In XOR, the output is 1 only when the inputs are different (0,1 or 1,0).
* If we plot the XOR inputs in a 2D space, we cannot draw a single straight line that separates the 1 outputs from the 0 outputs.
* A single-layer perceptron can only draw straight lines. XOR needs a curved or combined boundary, so a single layer fails.

**What this limitation tells us**

This shows that some problems require non-linear decision boundaries. To solve such problems, we need multilayer neural networks (also called multi-layer perceptrons or MLPs).
* By adding hidden layers and non-linear activation functions, we can combine simple linear boundaries to form complex, non-linear decision regions.
* This allows us to solve problems like XOR, AND, OR, and much more complex patterns.

# Question 06: [ Marks 20 ]

What is meant by the “perceptron Learning Rule” or weight adjustment method?
Why can adjusting one weight affect the classification of other points?

## Write Answer 06:

**Perceptron Learning Rule**

The perceptron learning rule is a method we use to adjust the weights and bias of a perceptron so that it can classify input data correctly.

* We start with initial weights and bias.
* For each training example, we calculate the perceptron's output.
* If the prediction is correct, we do nothing.
* If the prediction is wrong, we update the weights and bias using:

$$
w_i \gets w_i + \eta(y - \hat{y})x_i
$$
$$
b \gets b + \eta(y - \hat{y})
$$

Where:
* $w_i =$ weight of input $x_i$
* $\eta =$ learning rate
* $y =$ actual target output
* $\hat{y} =$ predicted output
* $b =$ bias

We repeat this process for all training examples until all points are classified correctly (if the data is linearly separable). The perceptron learning rule lets us correct mistakes by updating weights. Changing one weight moves the decision boundary, which can affect other points, so we repeat updates until all points are classified correctly.

**Why we adjusting one weight can affect other points?**

When we change a weight, it changes the slope or position of the decision boundary.
* This can cause points that were previously classified correctly to be misclassified.
* That's why updating weights affects the classification of all points, not just the one that caused the update.
* To fix this, we go through the dataset multiple times, adjusting weights until the perceptron correctly classifies all points.

# Question 07: [ Marks 20 ]

What are the main limitations of a single-layer perceptron?
How does the idea of using more than one layer (MLP) help overcome these limitations at a high level accoriding to you?



## Write Answer 07:

**Main Limitations of a Single-Layer Perceptron**

A single-layer perceptron has some important limitations:
1. **Can only solve linearly separable problems**: It can classify data only if a straight line (or hyperplane) can separate the classes. Problems like XOR, where the data is not linearly separable, cannot be solved.
2. Limited representation power: It cannot model complex patterns or relationships in data. It can only create simple linear decision boundaries. A single-layer perceptron can only draw straight lines, so it cannot solve complex or non-linear problems.
3. No feature learning: A single-layer perceptron cannot automatically extract or combine features from input data.

**How Multiple Layers Help**

By using more than one layer, also called a multi-layer perceptron (MLP):

* We can add hidden layers between the input and output layers.
* Hidden layers allow the network to combine linear boundaries to form non-linear decision boundaries.
* This lets the network learn complex patterns and solve problems that a single layer cannot, like XOR or real-world classification tasks.
* Using non-linear activation functions in hidden layers gives the network the ability to model almost any function, increasing its power and flexibility.
