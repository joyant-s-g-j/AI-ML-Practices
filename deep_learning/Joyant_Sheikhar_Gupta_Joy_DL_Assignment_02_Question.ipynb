# Question 01: [ Marks 10 ]
Explain gradient descent in detail. What is the main idea behind it? How does it help in minimizing the loss function? Explain the role of learning rate and describe what happens if the learning rate is too high or too low.

## Write Answer 01:
Gradient descent is an optimization algorithm used in deep learning to minimize the loss function of a model. The main idea is to adjust the model's parameters (weights and biases) step by step in the direction that reduces the loss, so the model can make better predictions.

**Main Idea**: The main ides of gradient descent is to minimize the loss function. The loss function tells us how far the model's predictions are from the real answers. By minimizing this loss, the model becomes more accurate.

**How it works**: Gradient descent starts with random values for the model's parameters, called weights. First, the model calculates the loss, which shows how far its predictions are from the actual answers, Then it figures out the gradient of the loss with respect to each weight. The gradient acts like a slope pointing in the direction of the steepest increase, so the model moves in the opposite direction to reduce the loss. The weights are updated by taking a step opposite to the gradient and this process is repeated many times until the loss becomes as small as possible.
Mathematically, updating a weight $w$ looks like this:

$$w = w - \eta \cdot \frac{\partial L}{\partial w}
$$
where:
* $\eta$ is the learing rate
* $\frac{\partial L}{\partial w}$ is the gradient of the loss function

**Role of Learning Rate**: The learing rate controls how big each step is when updating weights,
* If the learing rate is too high, the steps are too big, and the model can overshoot the minimum, making it unstable or never converging.
* If the learning rate is too low, the steps are very small, so learning becomes super slow and it can take forever to reach the minimum.

# Question 02: [ Marks 10 ]

What is forward propagation in an artificial neural network? Explain step by step how data moves from the input layer to the output layer.

## Write Answer 01:
Forward propagation is the process of sending input data through an artificial neural network (ANN) to get an output. It's called "forward" because the data moves from the input layer, through hidden layers, to the output layer. This is how the network makes predictions.

**Step by step process**:

**Input Layer**: The netwrok receives input data such as features of a dataset (e.g. height, weight, age) and these inputs are represented as numerical values that are passed to the first hidden layer.

**Weighted Sum**: Each input is multiplied by a weight that represents its importance, then all these values are summed together and a bias term is added to adjust the result.
$$
z = \sum_{i=1}^{n} (x_i \cdot w_i) + b
$$
Where, $x_i$ is the input, $w_i$ is the weight, and $b$ is the bias.

**Activation Function**: The calculated value $z$ is passed through an activation function as ReLU, Sigmoid, or Tanh, which introduces non-linearity and allows the neural network to learn complex patterns.
$$
a = activation(z)
$$

**Hidden Layers**: The output from the activation function becomes the input for the next layer, and the same process of weighted sum and activation continues through all hidden layers.

**Output Layer**: After passing through all layers, the final values reach the output layer where the network produces its prediction, such as a numeric value for regression or class probabilities for classification.

So basically forward propagation is how the network processes inputs to produce outputs using weights, biases and activation functions.

# Question 03: [ Marks 15 ]
Why do we need activation functions in neural networks? What would happen if we only used linear transformations without any activation function?

## Write Answer 03:
Activation functions are important because they decide whether a neuron should be activated or not. In simpler words, they help the network learn complex patterns from data. Without activation functions, the network would just be a bunch of linear calculations, and it wouldn't be able to solve problems like recognizing images, speech or any non-linear data.


If we didn't use activation functions and only applied linear transformations, like $y = Wx + b$, the neural network would lose its ability to learn complex patterns. No matter how many layers we added, the entire network would still behave like a single linear function, because combining multiple linear functions results in another linear function. This means the network would not be able to capture non-linear relationships in the data. For instance, it would fail to separate data arranged in a circle, solve XOR problems, or handle any other complex patterns. As a result, such a network would be practically useless for most real-world task, since real-world data is usually non-linear and requires the network to model intricate relationships.

# Question 04: [ Marks 15 ]

Explain the sigmoid activation function. Where is it used? What are its advantages and disadvantages? What problem does it cause during training?

## Write Answer 04:

The sigmoid activation function is a mathematical function that takes any real number as input and outputs a value between 0 and 1. It looks like an "S-shaped" curve. The formula for sigmoid is:
$$\sigma(x) = \frac{1}{1+e^{-x}}$$

Uses of Sigmoid: Sigmoid is mainly used in the output layer of neural networks when we want to predict probabilities, such as in binary classification(yes/no, true/false).

Advatages of Sigmoid:
1. Outputs are between 0 and 1, which can be interpreted as probabilities.
2. Smooth and differentiable, which means it works well with gradient-based optimization like gradient descent.
3. It has a nice, simple mathematical formula and is easy to implement.

Disadvantages of Sigmoid:
1. When the input is very large or very small, the gradient becomes almost 0. This slows down learning in deep networks.
2. Outputs are not zero-centered, which can make optimization harder because gradients may always move in the same direction.
3. Computationally expensive compared to simpler functions like ReLU.

Problem it causes during training:
* In deep networks, if the gradients are too small, the weights stop updating effectively, and the network learns very slowly or sometimes not at all.
* This makes sigmoid less suitable for hidden layers in deep neural networks, through it is still fine for output layers in binary classification tasks

# Question 05: [ Marks 10 ]
Explain the softmax activation function in simple words. Why is it used in multi-class classification problems? How is it different from sigmoid?

## Write Answer 05:

The softmax activation function is used in neural networks to convert a set of numbers (like the outputs of the last layer) intto probabilities that all add up to 1. The formul for softmax is:
$$
\sigma(z_i) = \frac{e^{z_i}}{\sum_je^{z_j}}
$$
Where $z_i$ is the score for class $i$ and the sum in the denominator goes over all classes.

Why is it used in multi-class classification:
* In multi-class problems, there are more than two possible classes (like classifying animals into cat, dog or rabbit)
* Softmax turns the network outputs into probabilities for each class.
* The class with the highest probability is choosen as the prediction.

Difference from sigmoid:
1. Sigmoid is used for binary classification, it gives a probability for one class, between 0 and 1.
2. Softmax is used for multi-class classification, it gives probabilities for all classes together, and the sum of all probabilities is 1.
3. Sigmoid works independently on each output, while softmax considers all outputs together.

# Question 06: [ Marks 20 ]

Explain the structure of an Artificial Neural Network in detail. Describe the role of the input layer, hidden layers, and output layer. How do weights, biases, and activation functions work together to transform input data into predictions? Illustrate your explanation with a simple example.

## Write Answer 06:

An Artificial Neural Network(ANN) is a computing system inspired by the human brain. It consists of layers of nodes(neurons) that work together to process information and make predictions. The basic structure of an ANN includes three main types of layers: input layer, hidden layers and output layers.

1. Input Layer: The input layer is where the network receives raw data. Each neuron in this layer represents a feature of the data. Example: If we are predicting house prices, the input layer could have neurons for size, number of rooms, and location.

2. Hidden Layers: Hidded layers are the intermediate layers between input and output. They are called "hidden" because we don't directly see their outputs. Each neuron in a hidden layer performs a weighted sum of its inputs, adds a bias, and passes the results through an activation function. Hidden layers help the network learn complex patterns in the data.

3. Output layer: The output layer produces the final prediction. For regression, it might output a number (e.g., predicted price). For classification, it might output probabilties(using sigmoid for binary or softmax for multi-class)

How weights, biases, and activation functions work together:
1. Weights($W$) - They decide the importance of each input, Higher weight means that input has more influence on the neuron's output.
2. Bias($b$) - It allows the neurons to shift the activation function and adds flexibility to the model.
3. Activation Function($f$) - It introduces non-linearity, allowing the network to learn complex patterns.
The computation inside a neuron looks like this:
$$
z = \sum^n_{i=1}(x_i ⋅ w_i) + b
$$
$$
a = f(z)
$$
Where $x_i$ is the input, $w_i$ is the weight, $b$ is the biasc, and $f$ is the activation function.

Simple Example: Suppose we want to predict whether a student will pass or fail an exam based on hours studied and hours slept:

* Input layer: 2 neurons(hours studied, hours slept)
* Hidden layer: 2 neurons with ReLU activation
* Output layer: 1 neuron with sigmoid activation (probability of passing)
1. Inputs are multiplied by weights and added with biases.
2. Hidden neurons apply the ReLU activation function to learn patterns like "studying more helps, but too little sleep reduces performance"
3. The output neuron applies sigmoid to produce a probability between 0 and 1.

By adjusting weights and biases training, the network learns the best way to combine inputs to make accurate predictions.
